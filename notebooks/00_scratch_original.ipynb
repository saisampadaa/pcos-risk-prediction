```python
data = data2.copy()
```

```python
data.drop(
    columns=['Sl. No', 'Patient File No.'],
    inplace=True
)
```

```python
data.shape
```

```python
missing = data.isnull().sum()
missing[missing > 0]
```

```python
data['Marraige Status (Yrs)'] = data['Marraige Status (Yrs)'].fillna(
    data['Marraige Status (Yrs)'].median()
)
```

```python
data.select_dtypes(include='object').columns
```

```python
data['AMH(ng/mL)'] = pd.to_numeric(
    data['AMH(ng/mL)'],
    errors='coerce'
)

data['II    beta-HCG(mIU/mL)'] = pd.to_numeric(
    data['II    beta-HCG(mIU/mL)'],
    errors='coerce'
)
```

```python
data.isnull().sum().sort_values(ascending=False)
```

```python
for col in data.columns:
    if data[col].dtype in ['int64', 'float64']:
        data[col] = data[col].fillna(data[col].median())
```
